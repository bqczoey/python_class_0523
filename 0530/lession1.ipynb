```python
#資料結構
#程式結構
[1,2,3,4,5] #list,用逗號區隔，可以修改內容
scores=[90,80,70] #運算式 #保留了scores這個變數，裡面有三個元素，分別是90,80,70
print(scores) #印出scores裡面的內容 #敘述式
```

```python
#tuple,用括號區隔，建立後不可以修改內容 #read only
scores1=(90,80,70)
print(scores1)
print(scores1[0]) #印出scores1裡面的第一個元素
```

```python
#distionary,用大括號區隔，裡面是key:value的形式，key不能重複，value可以修改
students = {
    "Tom":90,
    "Mary":80,
    "John":70}
print(students['Tom']) #印出students裡面的Tom的分數
```

```python
scores=[90,80,70]
for s in scores: 
#for迴圈，s是變數，scores是list，會依序把scores裡面的key賦值給s，直到把scores裡面的元素都賦值給s為止
#for 自定變數 in 串列資料, map 資料
    print(s) #印出s的值
```

```python
#dict
students = {
	'chinese':89,
	'english':94,
	'math':65,
	'history':92,
	'discover': 95}

for s in students:#for in 只會抓key，不會抓key後面的value
	print(s, students[s])
	 
```
