```python
class Person:
    def __init__(self, name):
        # 建立一個 attribute
        self.name = name

    def introduce(self):
        # 注意：這個方法必須縮排，與 __init__ 對齊，表示它屬於 Person 類別
        print(f"Hi, my name is {self.name}")

# 以下所有的實例化與呼叫，都必須回到最左邊（不縮排），代表在類別外部執行
p2 = Person("robert")
print(p2.name)

p3 = Person("mary")
print(p3.name)

p4 = Person("zoey")
p4.introduce()
```
