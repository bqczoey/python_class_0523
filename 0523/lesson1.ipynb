```python
# while true 無線迴圈，直到使用者輸入正確的分數
# break 跳出迴圈;從break開始往下執行的都會跳出迴圈；所以while true一定會搭配break,否則會一直無限迴圈
# 學生總分為300
#有些學生可以加分5%

while True:
    try:
        scores = int(input("請輸入學生分數(最高300分):"))
        break
    except ValueError:
        print("輸入錯誤，請輸入整數！")

is_add = input("學生是否符合加分條件?(y,n)")

if is_add == "y": #單項選擇
	scores = scores * 1.05

# 四捨五入(5進位)
scores = int(scores + 0.5)

# 超過300分，上限取300
if scores > 300:
	scores = 300

print("學生分數:",scores)
```

```python
import random

# 產生 1~100 的隨機整數
answer = random.randint(1, 100)
#print(answer)
count = 0

print("=== 猜數字遊戲 ===")
print("請猜一個 1 到 100 的數字")

while True:
    guess = int(input("請輸入數字: "))
    
    count += 1

    if guess > answer:
        print("太大了！")
    elif guess < answer:
        print("太小了！")
    else:
        print(f"恭喜猜對了！答案是 {answer}")
        print(f"你總共猜了 {count} 次")
        break
```
